```python
# ---------------------------------------------------------------------------
# Note: This script is calculating the alpha-helix content of a given protein
#       structure.
# ---------------------------------------------------------------------------

# ---------------------------------------------------------------------------
# Configuration — edit these values to match your setup.
# ---------------------------------------------------------------------------

# Directory containing the toxin PDB files to analyze.
PDB_DIR = "/path_to_your_base_directory/pdb_files"
# ---------------------------------------------------------------------------
```

```python
import os
from Bio.PDB import PDBParser, DSSP
```

```python
# List of PDB codes to analyze
tox_list = ["2H5F", "1DTK", "4K09", "1ANS", "1VIB", "7OKF", "2I61", "2ASC", "2ATB"]
parser = PDBParser()
```

```python
for code in tox_list:
    structure_path = f"{PDB_DIR}/{code.lower()}.pdb"
    structure = parser.get_structure(code, structure_path)
    model = structure[0]

    dssp = DSSP(model, structure_path)

    total = len(dssp)
    helix = sum(1 for aa in dssp if aa[2] == "H")

    print(f"{code}: Alpha-helix content: {helix / total * 100:.2f} %")
```
